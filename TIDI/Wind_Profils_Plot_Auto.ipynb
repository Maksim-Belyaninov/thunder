```python
import xarray as xr
import pandas as pd
import numpy as np
import datetime as dt
import os
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap, Normalize
from matplotlib.dates import DateFormatter, MinuteLocator
from matplotlib.lines import Line2D
from matplotlib.patches import Rectangle
import cartopy.crs as ccrs
import cartopy.feature as cfeature
import warnings
warnings.filterwarnings('ignore', category=FutureWarning)

# ============= ОБЩИЕ НАСТРОЙКИ =============
file_year = 2018    # Год файла
satellite_dir = f'C:/Users/Maks/Desktop/Jupyter/satellite_data/TIDI/{file_year}'
hdf_matching_path = f'C:/Users/Maks/Desktop/Jupyter/output_full_data/{file_year}_satellite_overpass_matching.h5'
thunderbolts_path = f'C:/Users/Maks/Desktop/Jupyter/thunderbolts_data/hdf/{file_year}_thunderbolts_clastered.h5'
output_plot_dir = f'C:/Users/Maks/Desktop/Jupyter/wind_profiles/{file_year}'
os.makedirs(output_plot_dir, exist_ok=True)

# Загружаем данные о совпадениях
satellite_overpass_matching = pd.read_hdf(hdf_matching_path, key='matches')
print(f"Загружено совпадений: {len(satellite_overpass_matching)}")

# Загружаем данные о разрядах
thunderbolts_df = pd.read_hdf(thunderbolts_path, key='strikes')
print(f"Загружено разрядов: {len(thunderbolts_df)}")

# Функция для извлечения дня из имени файла
def extract_day_from_filename(filename):
    import re
    match = re.search(r'20\d\d(\d{3})', filename)
    if match:
        return int(match.group(1))
    return None

# Функция нормализации долготы
def normalize_lon_for_plot(lons):
    lons = np.array(lons)
    lons = np.where(lons > 180, lons - 360, lons)
    return lons

# ===== ЦИКЛ ПО ВСЕМ КЛАСТЕРАМ =====
for cluster_num in satellite_overpass_matching.index:
    print("\n" + "="*70)
    print(f"ОБРАБОТКА КЛАСТЕРА #{cluster_num}")
    print("="*70)
    
    try:
        # Получаем данные для кластера
        data_row = satellite_overpass_matching.loc[cluster_num]
        
        # Извлекаем имя файла TIDI 
        tidi_file_full = data_row['satellite_file_name']
        tidi_file = tidi_file_full.replace('.VEC', '')
        
        # Извлекаем день года из имени файла
        day = extract_day_from_filename(tidi_file)
        if day is None:
            print(f"  Не удалось извлечь день из имени файла {tidi_file}, пропускаем")
            continue
            
        # Временные метки
        matched_time = pd.to_datetime(data_row['matched_time'])
        last_flash_time = pd.to_datetime(data_row['flash_time'])
        
        print(f"  Файл: {tidi_file}")
        print(f"  День: {day}")
        print(f"  Время совпадения: {matched_time}")
        print(f"  Время разряда: {last_flash_time}")
        
        # ===== РАЗРЯДЫ КЛАСТЕРА =====
        cluster_thunderbolts = thunderbolts_df[thunderbolts_df['clnb'] == cluster_num]
        print(f"  Разрядов в кластере: {len(cluster_thunderbolts)}")
        
        # Находим основной разряд
        flash_matches = cluster_thunderbolts[cluster_thunderbolts.index == last_flash_time]
        
        if len(flash_matches) > 0:
            if len(flash_matches) == 1:
                thunderbolt = flash_matches.iloc[0]
                flash_lat = thunderbolt['lat']
                flash_lon = thunderbolt['lon']
                flash_amp_value = thunderbolt['amp']
            else:
                # Если несколько с одним временем, берем с максимальной амплитудой
                max_amp_idx = flash_matches['amp'].abs().idxmax()
                flash_lat = flash_matches.loc[max_amp_idx, 'lat']
                flash_lon = flash_matches.loc[max_amp_idx, 'lon']
                flash_amp_value = flash_matches.loc[max_amp_idx, 'amp']
                
                if isinstance(flash_lat, pd.Series):
                    flash_lat = flash_lat.iloc[0]
                    flash_lon = flash_lon.iloc[0]
                    flash_amp_value = flash_amp_value.iloc[0]
            
            print(f"  Найден разряд: координаты ({flash_lat:.3f}°, {flash_lon:.3f}°), мощность {abs(flash_amp_value):.0f} A")
        else:
            print(f"  ВНИМАНИЕ: Разряд не найден!")
            flash_lat, flash_lon, flash_amp_value = None, None, None
        
        # Определяем пути к файлам
        filtered_satellite_path = f'C:/Users/Maks/Desktop/Jupyter/output_filtered_satellite_data/{file_year}/trimmed_{tidi_file}_clnb_{cluster_num}.h5'
        tidi_path = os.path.join(satellite_dir, tidi_file + '.VEC')
        
        # Проверяем существование файлов
        if not os.path.exists(filtered_satellite_path):
            print(f"  Файл траектории не найден: {filtered_satellite_path}")
            continue
        if not os.path.exists(tidi_path):
            print(f"  Файл TIDI не найден: {tidi_path}")
            continue
        
        # Определяем временное окно
        time_window_start = matched_time - dt.timedelta(minutes=15)
        time_window_end = matched_time + dt.timedelta(minutes=15)
        
        # Разряды за 15 минут до попадания
        flash_times_15min = cluster_thunderbolts[
            (cluster_thunderbolts.index >= time_window_start) & 
            (cluster_thunderbolts.index < last_flash_time)
        ]
        print(f"  Разрядов за 15 мин до: {len(flash_times_15min)}")
        
        # ===== ЗАГРУЗКА ТРАЕКТОРИИ =====
        filtered_satellite_data = pd.read_hdf(filtered_satellite_path, key='satellite_data')
        filtered_satellite_data = filtered_satellite_data.reset_index().rename(columns={'index': 'date'})
        if not pd.api.types.is_datetime64_any_dtype(filtered_satellite_data['date']):
            filtered_satellite_data['date'] = pd.to_datetime(filtered_satellite_data['date'])
        
        satellite_df = filtered_satellite_data.copy()
        satellite_df['time_diff'] = abs(satellite_df['date'] - matched_time)
        match_point = satellite_df.loc[satellite_df['time_diff'].idxmin()].copy()
        
        # ===== ЗАГРУЗКА ДАННЫХ TIDI =====
        ds = xr.open_dataset(tidi_path, decode_timedelta=True)
        df_full = ds.to_dataframe().reset_index()
        ds.close()
        
        # Конвертация времени
        base_date = pd.Timestamp('1980-01-06 00:00:00')
        df_full['datetime'] = base_date + pd.to_timedelta(df_full['time'], unit='ns')
        
        # Находим центральное измерение
        df_full['time_diff'] = abs(df_full['datetime'] - matched_time)
        closest_row = df_full.loc[df_full['time_diff'].idxmin()]
        center_nvec = closest_row['nvec']
        center_datetime = closest_row['datetime']
        
        # Данные во временном окне
        data_in_window = df_full[
            (df_full['datetime'] >= time_window_start) & 
            (df_full['datetime'] <= time_window_end)
        ]
        nvecs_in_window = data_in_window['nvec'].unique()
        
        # ===== СБОР ПРОФИЛЕЙ =====
        all_profiles = []
        
        # Создаем словарь для быстрого поиска координат траектории по времени
        trajectory_coords = {}
        for _, row in satellite_df.iterrows():
            time_floor = row['date'].floor('1s')
            trajectory_coords[time_floor] = (row['lat'], normalize_lon_for_plot(row['lon']))
        
        for nvec in nvecs_in_window:
            nvec_data = df_full[df_full['nvec'] == nvec]
            for _, row in nvec_data.iterrows():
                if np.isnan(row['u']) or np.isnan(row['v']):
                    continue
                
                # Проверяем близость к траектории
                meas_time_floor = pd.to_datetime(row['datetime']).floor('1s')
                
                if meas_time_floor in trajectory_coords:
                    traj_lat, traj_lon = trajectory_coords[meas_time_floor]
                    meas_lat = row['lat']
                    meas_lon_norm = normalize_lon_for_plot(row['lon'])
                    
                    # Рассчитываем расстояние
                    lat_diff = abs(meas_lat - traj_lat)
                    lon_diff = abs(meas_lon_norm - traj_lon)
                    
                    # Если точка близко к траектории (в пределах 1 градуса)
                    if lat_diff <= 1.0 and lon_diff <= 1.0:
                        wind_speed = np.sqrt(row['u']**2 + row['v']**2)
                        all_profiles.append({
                            'datetime': row['datetime'],
                            'nvec': nvec,
                            'altitude_km': row['alt_retrieved'],
                            'wind_speed': wind_speed,
                            'lat': row['lat'],
                            'lon': row['lon']
                        })
        
        if not all_profiles:
            print(f"  Нет данных ветра для кластера #{cluster_num}")
            continue
            
        filtered_profiles_df = pd.DataFrame(all_profiles)
        filtered_profiles_df = filtered_profiles_df[
            (filtered_profiles_df['datetime'] >= time_window_start) & 
            (filtered_profiles_df['datetime'] <= time_window_end)
        ].sort_values('datetime')
        
        # ===== ТОЧКИ НА КАРТЕ =====
        unique_nvecs = filtered_profiles_df['nvec'].unique()
        
        profile_points = []
        for nvec in unique_nvecs:
            nvec_data = filtered_profiles_df[filtered_profiles_df['nvec'] == nvec]
            first_row = nvec_data.iloc[0]
            point_time = first_row['datetime']
            
            # Находим ближайшую точку траектории по времени
            time_diff = abs(satellite_df['date'] - point_time)
            closest_traj_point = satellite_df.loc[time_diff.idxmin()]
            
            # Рассчитываем расстояние до траектории
            point_lon_norm = normalize_lon_for_plot(first_row['lon'])
            traj_lon_norm = normalize_lon_for_plot(closest_traj_point['lon'])
            
            lat_diff = abs(first_row['lat'] - closest_traj_point['lat'])
            lon_diff = abs(point_lon_norm - traj_lon_norm)
            
            # Если точка близко к траектории (в пределах 1 градуса)
            if lat_diff <= 1.0 and lon_diff <= 1.0:
                time_from_match = (point_time - matched_time).total_seconds() / 60
                if abs(time_from_match) <= 15:
                    profile_points.append({
                        'lat': first_row['lat'],
                        'lon': first_row['lon'],
                        'time': point_time,
                        'nvec': nvec
                    })
        
        profile_points_df = pd.DataFrame(profile_points)

        # Нормализация долготы для точек измерений
        if len(profile_points_df) > 0:
            profile_points_df['lon_plot'] = normalize_lon_for_plot(profile_points_df['lon'])
            
            # === ВАЖНО: Сначала фильтруем профили по всем найденным nvec ===
            valid_nvecs_all = profile_points_df['nvec'].unique()
            filtered_profiles_df = filtered_profiles_df[filtered_profiles_df['nvec'].isin(valid_nvecs_all)]
            
            # === Теперь ищем совпадение ===
            has_coincidence = False
            profile_points_df['time_diff'] = abs(profile_points_df['time'] - matched_time)
            closest_meas_idx = profile_points_df['time_diff'].idxmin()
            closest_meas_time_diff = profile_points_df.loc[closest_meas_idx, 'time_diff']
            
            # Если разница меньше 30 секунд – считаем совпадением
            if closest_meas_time_diff < pd.Timedelta(seconds=30):
                has_coincidence = True
                # Запоминаем координаты, время и nvec совпадения
                coincidence_lon = profile_points_df.loc[closest_meas_idx, 'lon_plot']
                coincidence_lat = profile_points_df.loc[closest_meas_idx, 'lat']
                coincidence_time = profile_points_df.loc[closest_meas_idx, 'time']
                coincidence_nvec = profile_points_df.loc[closest_meas_idx, 'nvec']
                # Удаляем эту точку ТОЛЬКО из profile_points_df для отображения на карте
                profile_points_df = profile_points_df.drop(index=closest_meas_idx).reset_index(drop=True)
        
        if len(profile_points_df) == 0:
            print(f"  Нет точек на траектории для кластера #{cluster_num}")
            continue
        
        print(f"  Отобрано точек рядом с траекторией: {len(profile_points_df)}")
        if has_coincidence:
            print(f"  Найдено совпадение! nvec={coincidence_nvec}, время={coincidence_time}")
        
        # ===== НОРМАЛИЗАЦИЯ ДОЛГОТ =====
        satellite_df['lon_plot'] = normalize_lon_for_plot(satellite_df['lon'])
        profile_points_df['lon_plot'] = normalize_lon_for_plot(profile_points_df['lon'])
        match_point['lon_plot'] = normalize_lon_for_plot(match_point['lon'])
        
        # ===== ПОСТРОЕНИЕ ГРАФИКА =====
        fig = plt.figure(figsize=(24, 10))
        gs = fig.add_gridspec(1, 2, width_ratios=[0.6, 0.4], wspace=0.15)
        
        # Левый график - профили
        ax1 = plt.subplot(1, 2, 1)
        colors = [(0,0,0.5), (0,0,1), (0,0.5,1), (0.5,1,0.5), (1,1,0), (1,0.5,0), (1,0,0), (0.5,0,0)]
        wind_cmap = LinearSegmentedColormap.from_list("wind_speed", colors, N=100)
        
        scatter = ax1.scatter(filtered_profiles_df['datetime'], 
                            filtered_profiles_df['altitude_km'], 
                            c=filtered_profiles_df['wind_speed'], 
                            cmap=wind_cmap, s=60, alpha=0.9,
                            vmin=0, vmax=200, edgecolors='black', linewidth=0.5)
        
        ax1.axvline(x=last_flash_time, color='black', linestyle='-', linewidth=2, alpha=0.8)
        ax1.axvline(x=center_datetime, color='gray', linestyle='--', linewidth=2, alpha=0.8)
        
        ax1.xaxis.set_major_formatter(DateFormatter('%H:%M'))
        ax1.xaxis.set_major_locator(MinuteLocator(interval=5))
        ax1.tick_params(axis='x', rotation=45)
        
        ax1.set_xlabel('Время (UTC)', fontsize=12)
        ax1.set_ylabel('Высота (км)', fontsize=12)
        ax1.set_xlim(time_window_start, time_window_end)
        ax1.set_ylim(60, 120)
        ax1.set_yticks(np.arange(60, 121, 10))
        ax1.grid(True, alpha=0.3)
        ax1.set_title(f'Вертикальные профили ветра (кластер #{cluster_num})', fontsize=14, fontweight='bold')
        
        cbar = plt.colorbar(scatter, ax=ax1)
        cbar.set_label('Скорость ветра (м/с)', fontsize=12)
        
        legend_elements = [
            Line2D([0], [0], color='gray', linestyle='--', linewidth=2,
                   label=f'Попадание: {center_datetime.strftime("%H:%M:%S")}'),
            Line2D([0], [0], color='black', linestyle='-', linewidth=2,
                   label=f'Разряд: {last_flash_time.strftime("%H:%M:%S")}')
        ]
        ax1.legend(handles=legend_elements, loc='upper right', fontsize=10)
        
        # Правый график - карта
        cluster_bounds = {
            'lat_min': data_row['lat_min'], 'lat_max': data_row['lat_max'],
            'lon_min': data_row['lon_min'], 'lon_max': data_row['lon_max']
        }
        
        ax2 = fig.add_subplot(1, 2, 2, projection=ccrs.PlateCarree())
        ax2.set_title('Карта пролета', fontsize=14, fontweight='bold', pad=10)
        
        ax2.add_feature(cfeature.LAND, facecolor='#e0e0e0')
        ax2.add_feature(cfeature.OCEAN, facecolor='#c6e2ff')
        ax2.add_feature(cfeature.COASTLINE, linewidth=0.8, edgecolor='#555555')
        ax2.add_feature(cfeature.BORDERS, linewidth=0.5, edgecolor='#555555', linestyle=':')
        
        gl = ax2.gridlines(draw_labels=True, crs=ccrs.PlateCarree(), 
                          linewidth=0.5, color='gray', alpha=0.5, linestyle='--')
        gl.top_labels = False
        gl.right_labels = False
        
        # Границы с использованием lon_plot
        all_lons = np.concatenate([satellite_df['lon_plot'].values, 
                                  [cluster_bounds['lon_min'], cluster_bounds['lon_max']],
                                  profile_points_df['lon_plot'].values])
        all_lats = np.concatenate([satellite_df['lat'].values, 
                                  [cluster_bounds['lat_min'], cluster_bounds['lat_max']],
                                  profile_points_df['lat'].values])
        
        if flash_lon is not None:
            all_lons = np.concatenate([all_lons, [normalize_lon_for_plot(flash_lon)]])
            all_lats = np.concatenate([all_lats, [flash_lat]])
        
        lon_padding = 10
        lat_padding = 5
        
        ax2.set_xlim(all_lons.min() - lon_padding, all_lons.max() + lon_padding)
        ax2.set_ylim(all_lats.min() - lat_padding, all_lats.max() + lat_padding)
        ax2.set_xlabel('Долгота', fontsize=11)
        ax2.set_ylabel('Широта', fontsize=11)
        
        # Траектория с lon_plot
        lons_fixed = satellite_df['lon_plot'].values
        lons_fixed = np.where(lons_fixed < 0, lons_fixed + 360, lons_fixed)
        threshold = 180
        segments = []
        current_segment = []
        for i in range(len(lons_fixed)):
            if len(current_segment) == 0:
                current_segment.append(i)
            else:
                prev_idx = current_segment[-1]
                if abs(lons_fixed[i] - lons_fixed[prev_idx]) > threshold:
                    segments.append(current_segment)
                    current_segment = [i]
                else:
                    current_segment.append(i)
        if current_segment:
            segments.append(current_segment)
        
        for seg in segments:
            ax2.plot(satellite_df['lon_plot'].iloc[seg], satellite_df['lat'].iloc[seg], 
                    color='red', linewidth=2.5, label='Траектория' if seg is segments[0] else '',
                    alpha=0.9, transform=ccrs.PlateCarree(), zorder=3)
        
        # Разряды до (за 15 минут)
        if len(flash_times_15min) > 0:
            ax2.scatter(flash_times_15min['lon'], flash_times_15min['lat'], 
                       marker='^', color='lightgray', s=80, alpha=1,
                       label=f'Разряды 15мин до ({len(flash_times_15min)})',
                       transform=ccrs.PlateCarree(), zorder=5, 
                       edgecolor='gray', linewidth=0.3)
        
        # Точка разряда (основной)
        if flash_lon is not None and flash_lat is not None:
            flash_lon_norm = normalize_lon_for_plot(flash_lon)
            norm = Normalize(vmin=0, vmax=32000)
            colors = [(0.5,0.0,0.5), (0,0,1), (0,0.5,1), (0.5,1,0.5), 
                      (1,1,0), (1,0.5,0), (1,0,0), (0.5,0,0)]
            flash_cmap = LinearSegmentedColormap.from_list("flash_cmap", colors, N=100)
            flash_point = ax2.scatter([flash_lon_norm], [flash_lat], 
                                      c=[abs(flash_amp_value)], cmap=flash_cmap, 
                                      norm=norm, s=120, marker='o', 
                                      edgecolor=None, linewidth=1.5,  
                                      label=f'Разряд ({abs(flash_amp_value):.0f} A)',
                                      zorder=30, transform=ccrs.PlateCarree())  
        
        # Точки измерений с lon_plot
        if len(profile_points_df) > 0:
            ax2.scatter(profile_points_df['lon_plot'], profile_points_df['lat'], 
                       color='black', s=100, marker='o', edgecolors=None, linewidth=1.5,
                       label=f'Точки измерений ({len(profile_points_df)})', transform=ccrs.PlateCarree(), zorder=10)
        
        # Подписи времени с lon_plot
        for _, point in profile_points_df.iterrows():
            ax2.text(point['lon_plot'] + 1.5, point['lat'] + 1.0, 
                    point['time'].strftime('%H:%M'), fontsize=8, fontweight='bold',
                    transform=ccrs.PlateCarree(), zorder=20,
                    bbox=dict(boxstyle='round,pad=0.2', facecolor='white', alpha=0.8, edgecolor='black'))
                
        # Зона кластера
        rect = Rectangle(
            (cluster_bounds['lon_min'], cluster_bounds['lat_min']),
            cluster_bounds['lon_max'] - cluster_bounds['lon_min'],
            cluster_bounds['lat_max'] - cluster_bounds['lat_min'],
            facecolor='blue', alpha=0.25, edgecolor='darkblue', linewidth=2,
            label='Зона кластера', transform=ccrs.PlateCarree(), zorder=2
        )
        ax2.add_patch(rect)
        
        # Маркер спутника TIMED/TIDI
        if has_coincidence:
            ax2.plot(coincidence_lon, coincidence_lat,
                     'g*', markersize=22, markeredgecolor='yellow', markeredgewidth=2,
                     label='TIMED/TIDI (с измерением)', transform=ccrs.PlateCarree(), zorder=25)
        else:
            ax2.plot(match_point['lon_plot'], match_point['lat'],
                     'r*', markersize=20, markeredgecolor='yellow', markeredgewidth=1.5,
                     label='TIMED/TIDI', transform=ccrs.PlateCarree(), zorder=25)

        # Подпись времени для точки совпадения
        if has_coincidence:
            ax2.text(coincidence_lon + 1.5, coincidence_lat + 1.0, 
                    coincidence_time.strftime('%H:%M'), fontsize=8, fontweight='bold',
                    color='black',
                    transform=ccrs.PlateCarree(),
                    bbox=dict(boxstyle='round,pad=0.2', facecolor='white', alpha=0.8, edgecolor='green'),
                    zorder=30)
        
        # Colorbar для разряда
        if flash_lon is not None and flash_lat is not None:
            pos = ax2.get_position()
            cax = fig.add_axes([pos.x1 + 0.01, pos.y0, 0.015, pos.height])
            norm = Normalize(vmin=0, vmax=32000)
            colors = [(0.5,0.0,0.5), (0,0,1), (0,0.5,1), (0.5,1,0.5), 
                      (1,1,0), (1,0.5,0), (1,0,0), (0.5,0,0)]
            flash_cmap = LinearSegmentedColormap.from_list("flash_cmap", colors, N=100)
            dummy_scatter = ax2.scatter([], [], c=[], cmap=flash_cmap, norm=norm)
            cbar_flash = plt.colorbar(dummy_scatter, cax=cax)
            cbar_flash.set_label('Мощность (A)', fontsize=11)
            cbar_flash.set_ticks([0, 8000, 16000, 24000, 32000])
        
        # Время существования кластера
        if len(cluster_thunderbolts) > 0:
            cluster_start_time = cluster_thunderbolts.index.min()
            cluster_end_time = cluster_thunderbolts.index.max()
            cluster_time = f'{cluster_start_time.strftime("%H:%M:%S")} - {cluster_end_time.strftime("%H:%M:%S")}'
        else:
            cluster_time = 'нет данных'
        
        # Формируем информационный текст
        info_text = ''
        if 'flash_amp' in data_row and pd.notna(data_row['flash_amp']):
            info_text += f'Мощность разряда: {abs(data_row["flash_amp"]):.0f} A\n'
        info_text += f'Разрядов за 15 мин до: {len(flash_times_15min)}\n'
        info_text += f'Период кластера: {cluster_time}'
        
        # Добавляем текст под картой
        ax2.text(0.02, -0.4, info_text, transform=ax2.transAxes, fontsize=10,
                fontweight='bold', verticalalignment='top',
                bbox=dict(boxstyle='round,pad=0.5', facecolor='yellow', alpha=0.3))
        
        # Легенда снизу 
        handles, labels = ax2.get_legend_handles_labels()
        
        # Удаляем дубликаты (если есть)
        unique = dict(zip(labels, handles))
        handles = list(unique.values())
        labels = list(unique.keys())
        
        ax2.legend(handles, labels, loc='upper center', bbox_to_anchor=(0.5, -0.15),
                  fontsize=8, ncol=4, framealpha=1.0)
        
        # Общий заголовок
        date_str = center_datetime.strftime('%d.%m.%Y')
        fig.suptitle(f'TIDI - {tidi_file} | {date_str} | Кластер #{cluster_num}', 
                    fontsize=16, fontweight='bold', y=0.98)
        
        plt.tight_layout()
        plt.subplots_adjust(left=0.05, right=0.88, bottom=0.08, top=0.92, wspace=0.15)
        
        # Сохранение
        plot_path = os.path.join(output_plot_dir, 
                                f'TIDI_cluster{cluster_num}_{date_str.replace(".", "")}_{center_datetime.strftime("%H%M%S")}.png')
        plt.savefig(plot_path, dpi=300, bbox_inches='tight')
        plt.close(fig)
        
        print(f"  График сохранен: {plot_path}")
        
    except Exception as e:
        print(f"  ОШИБКА при обработке кластера #{cluster_num}: {e}")
        import traceback
        traceback.print_exc()
        continue

print("\n" + "="*70)
print("ОБРАБОТКА ЗАВЕРШЕНА")
print("="*70)
```
